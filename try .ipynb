```python
import yfinance as yf

df = yf.download(
    "AAPL",
    start="2024-01-01",
    end="2025-01-01",
    auto_adjust=False
)

df = df.reset_index()

print(df.head())
print("Column names:", df.columns.tolist())
print("Number of columns:", len(df.columns))
```

```python
# Flatten the MultiIndex columns
df.columns = df.columns.get_level_values(0)

# Display the dataset
print(df.head())
```
