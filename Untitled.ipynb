```python
import yfinance as yf

# Choose 10 stocks
stocks = ["^DJI"]

for stock in stocks:
    df = yf.download(stock, start="2016-01-01", end="2025-12-31")
    df.reset_index(inplace=True)   # make Date a normal column 
    df.to_csv(f"{stock}.csv", index=False)
    print(f"{stock} saved")

print("Download complete!")
```
